## 6. Feature Engineering 

Cette partie consiste à créer de nouvelles variables plus informatives.

### 0. Petit point méthode : 

La question n'est donc pas : **"Quelles variables puis-je créer ?"**

mais plutôt :**"Quelle information utile n'est pas encore représentée par mes variables actuelles ?"**

1. Le point de départ c'est l'EDA : 
Tu ne fais jamais du Feature Engineering avant d'avoir exploré les données. L'EDA te révèle par exemple :
-   des variables très corrélées ;
-   des distributions particulières ;
-   des interactions entre variables ;
-   des informations cachées dans une colonne.

C'est à partir de ces observations que tu proposes de nouvelles variables.

2. Les grandes familles de Feature Engineering
    a. Extraire une information cachée : Statut social par exemple
    b. Combiner plusieurs variables : Création de FamilySize
    c. Simplifier une variable : Regroupement sous Rare des statuts rares
    d. Transformer une variable numérique : par exemple au lieu de Salaire, on peut créer la variable Salaire élevé
    e. Créer des interactions : Parfois, une variable seule explique peu, mais deux variables ensemble sont très informatives : 
        Avec Sex et Pclass on peut créer par exemple Sex*Pclass

3. Quand créer une nouvelle variable 

    Pose toi ces questions : 

    a. Est-ce que cette nouvelle variable apporte une information nouvelle ?
    b. Est-ce qu'elle simplifie le problème ?
    c. Est-ce qu'elle traduit une connaissance métier ?

4. Quand ne pas faire de Feature Engineering 

    a. Si la variable est redondante : créer Prix TTC = HT + TVA si ces deux dernières sont connues ne sert à rien car cette infos est déjà disponible
    b. Si la variable est arbitraire : cela n'a pas de sens 
    c. Si cela introduit une fuite de données : Le client va-t-il quitter l'entreprise? et on souhaite créer : **Nombre de jours après la résiliation** ; pas de sens car on ne connait pas cela au moment de la prédiction. 
    d. Si cela va compliquer le modèle sans bénéfice : on ne crée pas 20 variables si 2 suffisent

5. Comment savoir si une variable est utile? 

Après l'avoir créée, tu peux :
-   refaire une petite analyse (distribution, relation avec la cible) ;
-   mesurer son importance avec un modèle (Random Forest, XGBoost) ;
-   comparer les performances du modèle avec et sans cette variable.

Une variable qui n'améliore rien peut être supprimée.



### 1. Extraction du titre de la personne

La variable Name contient une information sociale.

Exemple : Dans "Braund, Mr. Owen Harris", on extrait Mr


```python
#df["Title"] = df["Name"].str.extract(", ([A-Za-z]+)\.", expand=False)
```

```python
df["Title"] = (
    df["Name"]
    .str.extract(r",\s*([^.]*)\.")[0]
    .str.strip()
)

print(
    df["Title"].value_counts()
)

df["Title"] = df["Title"].replace({
    "Mlle": "Miss",
    "Ms": "Miss",
    "Mme": "Mrs"
})

common_titles = [
    "Miss",
    "Mr",
    "Mrs",
    "Master"
]

df["Title"] = df["Title"].where(
    df["Title"].isin(common_titles),
    "Rare"
)

print(
    df["Title"].value_counts()
)
```

On visualise les différentes valeurs obtenues : 

```python
df["Title"].value_counts()
```

Remarque : On a beaucoup trop de titre pour que cela soit exploitable, donc on regroupe les titres rares, sous une même case : 

```python
rare_titles = [
    "Lady","Countess", "the Countess","Capt",
    "Col","Don","Dr","Major",
    "Rev","Sir","Jonkheer"
]

df["Title"] = df["Title"].replace(
    rare_titles,
    "Rare"
)
```

On visualise à nouveau les différentes valeurs obtenues : 

```python
df["Title"].value_counts()
```

Avec le tableau suivant de correspondances : 

| Titre      | Signification       | Utilisation                            |
| ---------- | ------------------- | -------------------------------------- |
| **Mr**     | Monsieur            | Homme adulte                           |
| **Mrs**    | Madame              | Femme mariée                           |
| **Miss**   | Mademoiselle        | Femme non mariée (ou jeune fille)      |
| **Master** | Maître              | Jeune garçon                           |
| **Ms**     | Madame/Mademoiselle | Femme sans préciser son statut marital |


On peut remplacer Mlle par Miss, Ms avec Miss par choix (on aurait pu le remplacer par Mrs), Mme par Mrs : 

```python
df["Title"] = df["Title"].replace({
    "Mlle": "Miss",
    "Ms": "Miss",
    "Mme": "Mrs"
})
```

```python
# Vérification 
df["Title"].value_counts()
```

### 2. Variables à plusieurs catégories : Embarked, Title

On utilise One-Hot Encoding, mais le provlème à étudier est dans le paramètre drop_first. 

Expliquons ce que fait drop_first : 

Lorsqu'on applique un One-Hot Encoding, chaque catégorie est transformée en une colonne binaire.

Exemple avec **Embarked**:

| Embarked |
| -------- |
| C        |
| Q        |
| S        |

-   Sans drop_first : 

| Embarked_C | Embarked_Q | Embarked_S |
| ---------- | ---------- | ---------- |
| 1          | 0          | 0          |
| 0          | 1          | 0          |
| 0          | 0          | 1          |


-   Avec drop_first : 

| Embarked_Q | Embarked_S |
| ---------- | ---------- |
| 0          | 0          |
| 1          | 0          | 
| 0          | 1          | 

La catégorie C devient la catégorie de référence.

comme : Embarked_C + Embarked_Q + Embarked_S = 1 ; Les trois colonnes sont parfaitement dépendantes. Il s'agit de la multicolinéarité.

Quand utiliser le paramètre ? 

-   Pour des modèle qui estime des coefficients et que l'on souhaite éviter la multicolinéarité : Reg.Log., Reg.Lin., GLM. Dans ces modèles conserver toutes les colonnes peut rendre instable les coefficients ou difficile à interpréter.

Quand ne pas utiliser le paramètre ? 

-   Pour les modèles basés sur des arbres ou sur des règles de séparation :
    -   Decision Tree
    -   Random Forest
    -   XGBoost
    -   LightGBM
    -   CatBoost

Ces modèles ne sont pas affectés par la multicolinéarité des variables issues du One-Hot Encoding.

On va donc créer deux datasets issu du premier : un avec drop_first = false pour les modèle ne se souciant pas de la multicolinéarité; et True pour ceux qui peuvent en souffrir.








```python
# Création du 1er dataset où drop_first = False
df_full_columns = pd.get_dummies(
    df, columns=["Embarked", "Title"],
    drop_first=False, dtype=int)                            

# Création du 2ème dataset où drop_first = True
df_less_columns = pd.get_dummies(
    df, columns=["Embarked", "Title"],
    drop_first=True, dtype=int)    
```
